```python
# Create sample file.txt matching the PDF input
sample_content = """This is the first word.
This is the second text, Hello! How are you?
This is the third, this is it now."""

with open("file.txt", "w", encoding="utf-8") as f:
    f.write(sample_content)

with open("file.txt", "r", encoding="utf-8") as f:
    print("FILE CONTENT:\n" + f.read())
```

```python
import string
import nltk
from nltk.corpus import stopwords

# Download English stop words
nltk.download("stopwords", quiet=True)

# Step 1: Read the file
with open("file.txt", "r", encoding="utf-8") as file:
    lines = file.readlines()

print("Number of lines in file:", len(lines))

# Step 2: Remove punctuation and convert to lowercase
cleaned_lines = []
for line in lines:
    line = line.translate(
        str.maketrans("", "", string.punctuation)
    )
    line = line.lower()
    cleaned_lines.append(line)

print("\nCleaned lines:")
print(cleaned_lines)

# Step 3: Tokenize each line
tokenized_lines = []
for line in cleaned_lines:
    tokens = line.split()
    tokenized_lines.append(tokens)

print("\nTokenized data:")
print(tokenized_lines)

# Step 4: Remove stop words
stop_words = set(stopwords.words("english"))
tokens_without_sw = []
for tokens in tokenized_lines:
    for word in tokens:
        if word not in stop_words:
            tokens_without_sw.append(word)

print("\nTokens without stop words:")
print(tokens_without_sw)

# Step 5: Create the inverted index
inverted_index = {}
for line_number, tokens in enumerate(tokenized_lines, start=1):
    for word in tokens:
        if word not in stop_words:
            if word not in inverted_index:
                inverted_index[word] = []
            if line_number not in inverted_index[word]:
                inverted_index[word].append(line_number)

print("\nInverted Index:")
print(inverted_index)

# Step 6: Search for a word
search_word = input("\nEnter a word to search: ").lower()
search_word = search_word.translate(
    str.maketrans("", "", string.punctuation)
)

if search_word in inverted_index:
    print("Word found in line(s):", inverted_index[search_word])
else:
    print("Word not found.")
```
