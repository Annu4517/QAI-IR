```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
import string

# Download the English stop-word list
nltk.download("stopwords", quiet=True)

# Step 1: Input text
text = """
The students are studying information retrieval techniques.
They were searching for useful documents and learning concepts.
"""

print("Original Text:")
print(text)

# Step 2: Convert text to lowercase
text = text.lower()

# Step 3: Remove punctuation
text = text.translate(str.maketrans("", "", string.punctuation))

# Step 4: Tokenize the text into words
tokens = text.split()

print("\nTokens:")
print(tokens)

# Step 5: Remove stop words
stop_words = set(stopwords.words("english"))

filtered_words = [
    word for word in tokens if word not in stop_words
]

print("\nAfter Stop-word Removal:")
print(filtered_words)

# Step 6: Apply stemming
stemmer = PorterStemmer()

stemmed_words = [
    stemmer.stem(word) for word in filtered_words
]

print("\nAfter Stemming:")
print(stemmed_words)

# Step 7: Display the final processed text
processed_text = " ".join(stemmed_words)

print("\nFinal Preprocessed Text:")
print(processed_text)
```
